```python
import pandas as pd
import numpy as np
```

```python
# Task 1:

std_data = {
    "Name" : ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Marks": [70, 85, 60, 95, 45]
}

ds=pd.DataFrame(std_data)
ds
```

```python
marks = ds["Marks"]

#Average marks:
print("Average of Marks is: ")
print(marks.mean())

#Maximum marks:
print("Maximum value in Marks is: ")
print(marks.max())

#Minimum marks:
print("Minimum value in Marks is: ")
print(marks.min())

students = ds["Name"]
#Total Students:
print("Total Number of students are: ")
print(students.count())
```

```python
#Task 2:
passed_students = ds[ds["Marks"]>=50]
print("Passed Students are: ")
print(passed_students)

failed_students = ds[ds["Marks"]<=50]
print("Failed Students are: ")
print(failed_students)
```

```python
#Task 4:
std_info = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Courses": ["AI", "Web", "AI", "Web", "AI"],
    "Marks": [70, 85, 60, 95, 45]
}

ds = pd.DataFrame(std_info)
ds
```

```python
# Average Marks of each course

Avg_marks = ds.groupby("Courses")["Marks"].mean()
print("Average Marks of each course are: ")

print(Avg_marks)
```
